# Validação das bases

As bases preservam suas chaves únicas? Confere pedidos e itens processados.

```python
import pandas as pd
```

```python
from pathlib import Path

# Procura a raiz a partir da pasta atual; funciona também nas cópias em outputs/notebooks.
RAIZ_PROJETO = next(
    (pasta for pasta in [Path.cwd().resolve(), *Path.cwd().resolve().parents]
     if (pasta / 'executar_pipeline.py').is_file() and (pasta / 'src').is_dir()),
    None
)
if RAIZ_PROJETO is None:
    raise FileNotFoundError('Execute o notebook dentro da pasta do projeto ou de uma subpasta.')

CAMINHO_PROCESSADOS = RAIZ_PROJETO / 'data' / 'processed'

df_base_teste = pd.read_parquet(
    f'{CAMINHO_PROCESSADOS}/base_analitica_pedidos.parquet'
)

df_itens_teste = pd.read_parquet(
    f'{CAMINHO_PROCESSADOS}/itens_pedidos.parquet'
)
```

```python
assert df_base_teste['order_id'].is_unique
assert df_base_teste['order_id'].notna().all()

assert (
    df_itens_teste[
        ['order_id', 'order_item_id']
    ]
    .duplicated()
    .sum() == 0
)

assert df_itens_teste['order_id'].notna().all()
assert df_itens_teste['order_item_id'].notna().all()

print('Base analítica:', df_base_teste.shape)
print('Base de itens:', df_itens_teste.shape)

display(df_base_teste.head())
display(df_itens_teste.head())
```
